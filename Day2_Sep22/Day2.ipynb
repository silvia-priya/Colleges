### List comprehension vs loops


```python
#Loop to print the square of elements from 1 to 5
squares=[]
for i in range(1,5):
    squares.append(i**2)
print(squares)
```

```python
#Loop to print the square of elements from 1 to 5 using for loop
squares=[i**2 for i in range(1,5)]
print(squares)
```

```python
#Loop to print the square of elements from 1 to 10, but only for even numbers
squares_even=[i**2 for i in range(1,10) if i%2==0]
print(squares_even)
```

### Find whether a number is prime or not

```python
num=int(input("Enter the number"))
#False - means the number is not a prime number
#True - means the number is a prime number

def is_prime(num):
    if num<=1:
        return False
    for i in range(2,num):
        if num%i==0:
            return False
    return True


if is_prime(num):
    print("{} is  a prime number".format(num))
else:
    print("{} is not a prime number".format(num))




```
